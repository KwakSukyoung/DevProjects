```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
```

***Segment Guests***
- In this session, we will segment and analyze guests so we can answer this question "Are there distinct types of guests?" by using K-Mean Clustering

```python
data = pd.read_csv("../data/guest_survey_data.csv")
data
```

```python
features= data[["age","family_size","spending","satisfaction","learning_score","participated_booth"]]
scaler = StandardScaler()
X = scaler.fit_transform(features)
# Try 4 groups
kmeans = KMeans(n_clusters=4,random_state=42,n_init=10)

data["segment"] = kmeans.fit_predict(X)

# Now Profile
segment_profile = (data.groupby("segment")[["age",
                                            "family_size",
                                            "spending",
                                            "satisfaction",
                                            "learning_score",
                                            "participated_booth"]].mean())

segment_profile

```
