***Q2. What predicts whether guests are likely to return?***

```python
import numpy as np
import pandas as pd
import statsmodels.formula.api as smf
```

```python
data = pd.read_csv("../data/guest_survey_data.csv")
data
```

```python
# Create a binary variable:
data['likely_return'] = (data['likelihood_return'] >=4).astype(int)

# Apply logistic regression
model = smf.logit(
    """likely_return ~ satisfaction
    + learning_score
    + participated_booth
    + spending
    + C(education_program)
    """,
    data=data
).fit()
print(model.summary())
```

- Satisfaction: coef(3.343) - p-val(0.426) = Strong positive association
- learning_score: coef(-0.1054) - p-val(0.001) = No statistically significant association
- participated_booth: coef(-0.076) - p-val(0.347) = No statistically significant association
- spending: coef(0.012) - p-val(0.142) = Positive, but not statistically significant
- education_program(yes): coef(0.123) - p-val(0.608) = No statistically significant association

Conclusion: The logistic regression indicated that guest satisfaction was strongly and positively associated with return intent (β = 3.34, p < 0.001). Learning score, booth participation, spending, and education-program participation were not statistically significant predictors of return intent (p > 0.05). This suggests that overall guest satisfaction may be a more important driver of return intent than the individual experience and spending variables included in the model.
