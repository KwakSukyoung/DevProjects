***Qualitative Research***
- We will categorize and analyze the guest comments.

```python
import numpy as np
import pandas as pd
import json
from collections import Counter
# import statsmodels.models as smf
```

```python
with open("../data/guest_comments.json", "r", encoding="utf-8") as f:
    guest_comments = json.load(f)
print(guest_comments[0]["comment"])
```

```python
## Categorize the themes
themes = {"Satisfaction": ["satisfied","happy","enjoyed","great experience","wonderful","excellent","positive experience","fun","amazing","disappointed"],
          "Learning": ["learned","learning","educational","knowledge","understand","taught","discovered","informative","new things","science"],
          "Booth Participation": ["booth","activity booth","participated","participate","hands-on activity","activity","interactive activity","booth activity","participated in the booth","skipped the booth"],
          "Spending": ["spent","spending","money","expensive","price","cost","admission price","food","snacks","gift shop"],
          "Education Program": ["education program","educational program","program","class","lesson","instructor","educator","educational activity","education","learning program"],
          "Return Intent": ["come back","return","visit again","next visit","future visit","come again","would return","visit in the future","come back again","recommend"]}
```

```python
# Simple theme classifier
def identify_theme (comments, themes):
    for i in range(len(guest_comments)):
        comment = comments[i]["comment"].lower()

        matched_themes = []
        
        for theme, keywords in themes.items():
            for keyword in keywords:
                if keyword in comment:
                    matched_themes.append(theme)
                    break
    return matched_themes

themse_list = identify_theme(guest_comments, themes)
for comment in guest_comments: print(comment["comment"], "->", themse_list)
```
